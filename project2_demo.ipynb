There will be three agents in the workflow: ScrapingAgent, DataAnalysisAgent, and mainAgent.
quizUrl => mainAgent => ScrapingAgent => DataAnalysisAgent => mainAgent => finalOutput
ScrapingAgent: This agent is responsible for scraping quiz data from the provided URL. It will extract questions, and all the file conents to solve the quiz.
DataAnalysisAgent: This agent will analyze the scraped data to find patterns, answers, or any relevant information needed to solve the quiz.
mainAgent: This agent will coordinate the workflow, passing data between the ScrapingAgent and DataAnalysisAgent, and finally compiling the results into a final output.
